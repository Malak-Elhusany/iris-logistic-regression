```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
```

```python
df= pd.read_csv("iris.csv")
df.head()
```

```python
df.info()
```

## Preprocessing


```python
df.isna().sum()  #No missing values
```

```python
df=df.drop(columns=["Id"])
```

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
df["Species"]=le.fit_transform(df["Species"])
df.head(10)
```

```python
x=df.drop(columns=["Species"])
y=df["Species"]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)

print("x train shape ",x_train.shape)
print("y train shape ",y_train.shape)
print("x test shape ",x_test.shape)
print("y test shape ",y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler   #just for features data
sd=StandardScaler()
x_train_sc=sd.fit_transform(x_train)
x_test_sc=sd.transform(x_test)

print("x_train Scaled ",x_train_sc)
print("x_test Scaled ",x_test_sc)
```

## Logistic Regression Model

```python
from sklearn.linear_model import LogisticRegression
lg=LogisticRegression(random_state=42)
```

```python
lg.fit(x_train_sc,y_train)
```

```python
y_pred= lg.predict(x_test_sc)
```

```python
from sklearn.metrics import accuracy_score,classification_report

print("Accuracy: ",accuracy_score(y_test,y_pred))
print("\nClassification report: \n", classification_report(y_test,y_pred))
```

## Prediction on new data

```python
new_sample=np.array([[5.1,3.5,1.4,0.2]])
new_sample_sc=sd.transform(new_sample)
prediction=lg.predict(new_sample_sc)
prob=lg.predict_proba(new_sample_sc)

print("Prediction class: ",prediction[0])
print("Probability class: ",prob[0])
```
